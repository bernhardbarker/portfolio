## Nonlinear Programming: Portfolio Optimization

Suppose we have some budget $B$ along with a list of stocks with price $p_i$, expected future price $u_i$ and variance of future price $\sigma_i^2$. Decide whether and how much we should invest in each stock to minimize variance (risk) of total revenue, while ensuring the expected revenue is above some threshold $R$. This can be formulated as follows:

$$
\begin{aligned}
\max\quad & \sum_{i=1}^{n} \sigma_i^2 x_i^2 \\
\text{s.t.}\quad & \sum_{i=1}^{n} p_i x_i \le B \\
& \sum_{i=1}^{n} u_i x_i \ge R \\
& x_i \ge 0 \quad \forall i = 1,2,\dots,n
\end{aligned}
$$

Problem described in 4.11 of [Operations Research (2): Optimization Algorithms](https://www.coursera.org/learn/operations-research-algorithms/home/welcome).

```python
# Imports and input data
import numpy as np
import pandas as pd
from scipy.optimize import minimize, LinearConstraint

data = pd.DataFrame([
    (50, 55, 100),
    (40, 50, 1600),
    (25, 20, 100),
], columns=["price", "expected_price", "price_variance"])

budget = 100000
minimum_expected_revenue = 115000

data
```

```python
# Define objective function and constraints

def min_function(x):
    # Find minimum of sum( σ^2.xi^2 )
    return (data.price_variance * x ** 2).sum()

# n variables >= 0
bounds = [(0, None)] * len(data)

# Set budget and target revenue
constraints = [
    LinearConstraint(data["price"], ub=budget),
    LinearConstraint(data["expected_price"], lb=minimum_expected_revenue),
]
```

```python
# Run optimizer
result = minimize(min_function, x0=np.ones(len(data)), bounds=bounds, constraints=constraints)
result
```

```python
# Print solution details

def print_label_value(label, value):
    return print(f"{label:22s}", value)

print("Optimal solution:")
print_label_value("Objective value:", f"{result.fun:.2f}")
print_label_value("Spending:", f"{sum(result.x * data.price):.2f}")
print_label_value("Expected revenue:", f"{sum(result.x * data.expected_price):.2f}")
print_label_value("Quantity to purchase:", ", ".join(f"x{i} = {result.x[i]:.2f}" for i in range(len(data))))
```
