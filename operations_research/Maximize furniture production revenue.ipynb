## Integer Programming: Maximize Furniture Production Revenue

A factory produces chairs and tables.

- Each chair requires 3 units of wood, 1 hour of labor and 50 minutes of machine time.
- Each table requires 5 units of wood, 2 hours of labor and 20 minutes of machine time.

We have a supply of 3600 units of wood, 200 workers that can work 8 hours each (1600 hours in total) and 50 machines that each run for 16 hours (48000 minutes in total).

Chairs and tables are sold at $700 and $900 respectively.

This can be formulated as follow:

$$
\begin{array}{rlll}
\max\quad & 700x_1 + 900x_2 \\
\text{s.t.}\quad & 3x_1 + 5x_2 & \le 3600 & \text{(wood)}\\
& x_1 + 2x_2 & \le 1600 & \text{(labor)} \\
& 50x_1 + 20x_2 & \le 48000 & \text{(machine)} \\
& x_i \ge 0 \quad \forall i = 1,2
\end{array}
$$

Problem described in 2.7 of [Operations Research (1): Models and Applications](https://www.coursera.org/learn/operations-research-modeling/home/welcome).


```python
# Imports
import numpy as np
from scipy.optimize import LinearConstraint, milp
```

```python
# Set objective and constraints

# Flip coef sign to turn maximization problem into minimization problem
coef = -np.array([700, 900])

A = [[3, 5],
     [1, 2],
     [50, 20]]
b_ub = [3600, 1600, 48000]

constraint = LinearConstraint(A, ub=b_ub)
```

```python
# Run solver

# Set all variables to be integers
integrality = np.ones_like(coef)

res = milp(coef, integrality=integrality, constraints=constraint)
res
```

```python
print(f"Optimal solution: Produce {res.x[0]:g} chairs and {res.x[1]:g} tables for a profit of {-res.fun:g}")
```
