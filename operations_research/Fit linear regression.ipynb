## Nonlinear Programming: Fit Linear Regression
Given a list of points, find $a$ and $b$ such that the straight line $y = a + bx$ minimizes the sum of squared errors for all the data points, i.e. solve:

$$\min_{\alpha,\beta}\sum_{i=1}^{n}\left[ y_{i} - \left( \alpha + \beta x_{i} \right) \right]^{2}$$

Final problem (module 6 quiz) for [Operations Research: Models and Applications](https://www.coursera.org/learn/operations-research-modeling/home/welcome).


```python
# Input data
xys = [
    (38, 137),
    (56, 201),
    (50, 152),
    (52, 107),
    (37, 150),
    (60, 173),
    (67, 194),
    (54, 166),
    (59, 154),
    (43, 137),
    (30, 38),
    (53, 193),
    (59, 154),
    (40, 175),
    (65, 247),
]
```

```python
# Run optimizer
import matplotlib.pyplot as plt
import numpy as np
from scipy import optimize

xs, ys = np.transpose(np.array(xys))

def min_function(ab):
    # Find minimum of sum( (yi - (a + b.xi))^2 )
    return ((ys - (ab[0] + ab[1] * xs)) ** 2).sum()

# 2 unbounded variables a and b
bounds = [(None, None)] * 2

result = optimize.shgo(min_function, bounds)
result
```

```python
# Print solution details
optimal_ab = result.x
print(f"Optimal solution: a = {optimal_ab[0]:.2f}, b = {optimal_ab[1]:.2f}")
```

```python
# Plot data points with fitted line

# Plot data points
plt.scatter(xs, ys)

# Plot fitted line
x = np.array([min(xs), max(xs)])
plt.plot(x, optimal_ab[0] + optimal_ab[1] * x)

plt.title(f"Fitted line y = {optimal_ab[0]:.2f} + {optimal_ab[1]:.2f}*x")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```
