## Solve linear programming

Find an optimal solution to the following:

$$
\begin{aligned}
\max\quad & x_1 + 3x_2 \\
\text{s.t.}\quad & -x_1 + x_2 \le 3 \\
& -x_1 + 2x_2 \le 8 \\
& 3x_1 + x_2 \le 18 \\
& x_i \ge 0 \quad \forall i = 1,2
\end{aligned}
$$

Problem 1 of week 1 quiz from [Operations Research (3): Theory](https://www.coursera.org/learn/operations-research-theory/home/welcome).

```python
# Imports and common function
import numpy as np
from scipy.optimize import linprog

def res_to_string(res):
    return f"x1 = {res.x[0]:g}, x2 = {res.x[1]:g}, objective value = {-res.fun:g}"
```

```python
A_ub = np.array([[-1, 1], [-1, 2], [3, 1]])
b_ub = np.array([3, 8, 18])

# Flip coef sign to turn maximization problem into minimization problem
coef = -np.array([1, 3])

# Run standard linear programming
bounds = [(0, None), (0, None)]
res = linprog(coef, A_ub=A_ub, b_ub=b_ub, bounds=bounds)
res
```

```python
print("Linear programming solution:", res_to_string(res))
```

## Manual branch-and-bound to solve integer programming

Find an integer solution to the following:

$$
\begin{aligned}
\max\quad & 3x_1 + 5x_2 \\
\text{s.t.}\quad & x_1 + x_2 \le 16 \\
& x_2 \le 7.5 \\
& x_i \in \mathbb{Z}_+ \quad \forall i = 1,2
\end{aligned}
$$

Problem 1-3 of week 3 quiz for [Operations Research (2): Optimization Algorithms](https://www.coursera.org/learn/operations-research-algorithms/home/welcome).


```python
A_ub = np.array([[1, 1]])
b_ub = np.array([16])

# Flip coef sign to turn maximization problem into minimization problem
coef = -np.array([3, 5])

# Run standard linear programming
bounds = [(0, None), (0, 7.5)]
res = linprog(coef, A_ub=A_ub, b_ub=b_ub, bounds=bounds)
print("Linear programming solution:", res_to_string(res))
```

```python
# x1 is not an integer (8.5), so run linear programming bounded below its floor and above it's ceiling, i.e. x1 <= 8 and x1 >= 9
bounds = [(0, 8), (0, 7.5)]
res1 = linprog(coef, A_ub=A_ub, b_ub=b_ub, bounds=bounds)
print("x1 <= 8:", res_to_string(res1))

bounds = [(9, None), (0, 7.5)]
res2 = linprog(coef, A_ub=A_ub, b_ub=b_ub, bounds=bounds)
print("x1 >= 9:", res_to_string(res2))
```

```python
# The 2nd solution is an integer solution and has a higher objective value than the 1st solution, so we don't need to branch on the 1st solution
print("Optimal integer solution:", res_to_string(res2))
```
